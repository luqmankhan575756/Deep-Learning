```python
import numpy as np 
```

```python
X = np.array([2, 3, 4])
w = np.array([0.2, 0.5, 0.3])
b = 0.1
```

```python
z = np.dot(X, w)+b
print("Weighted sum:", z)
```

```python
y_pre = 1/(1+np.exp(-z))
print("Predicted output:", y_pre)
```

```python
#Actual output
y = 1
```

```python
loss = (y - y_pre) ** 2
print("Loss:", loss)
```

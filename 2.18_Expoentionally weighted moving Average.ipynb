```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("D:\Deep Learning\DailyDelhiClimate.csv")
df.head()
```

```python
df= df[['date', 'meantemp']]
df 
```

```python
plt.scatter(df['date'], df['meantemp'], color = 'red')
```

```python
xl = df['meantemp'].ewm(alpha=0.9).mean()
xl
```

```python
df['ewm']= xl
df 
```

```python
plt.scatter(df['date'], df['meantemp'], color = 'red')
plt.plot(df['date'],xl,color= 'black', label= 'alpha = 0.9')
plt.legend()
plt.show()
```

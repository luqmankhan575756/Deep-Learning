```python
datagen = ImageDataGenerator(
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='constant'
    
)
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
i = 0
for batch in datagen.flow(input_batch, batch_size=1, save_to_dir="D:\Deep Learning\ca vs dog dataset\Augmentation"):
    i += 1
    if i == 10:
        break
```
